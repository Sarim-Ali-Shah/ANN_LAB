## Task 3: Addition & Subtraction

```python
import numpy as np

term1_marks = np.array([
    [70, 80, 75],
    [60, 65, 70],
    [90, 85, 92]
])

term2_marks = np.array([
    [75, 82, 78],
    [62, 68, 73],
    [88, 87, 95]
])

print(term1_marks)
print("\n")
print(term2_marks)
```

```python
combined_marks = term1_marks + term2_marks
print("Total marks: \n", combined_marks)
```

```python
improvement_matrix = term2_marks - term1_marks
print("Improvement:\n", improvement_matrix)
```

```python
# each row is a student ( axis = 1)
total_improvement_per_student = np.sum(improvement_matrix, axis=1)

highest_improvement_student_index = np.argmax(total_improvement_per_student)


student_number = highest_improvement_student_index

print("Total improvement per student:", total_improvement_per_student)
print("Highest total improvement student number", student_number)
```

## Task 6: Indexing

```python
random_matrix = np.random.randint(1, 51, size=(5, 5))
print("Original 5x5 Random Matrix:\n", random_matrix)
```

```python
element = random_matrix[2, 3]
print("\nElement in 3rd row, 4th column:", element)
```

```python
last_row = random_matrix[-1 , :]
print("\nLast row using negative indexing:", last_row)

last_col = random_matrix[:, -1]
print("Last column using negative indexing:\n", last_col)
```

```python
random_matrix[0, 0] = 999
print("\nUpdated Matrix: ", random_matrix)
```

## Task 7: Slicing

```python
matrix = np.arange(1, 37).reshape(6, 6)
print("Original 6x6 Matrix:\n", matrix)
```

```python
top_left_submatrix = matrix[:3 , :3]
print("\nTop-left 3x3 Sub-matrix:\n", top_left_submatrix)
```

```python
alternate_rows_cols = matrix[::2 , ::2]
print("\nAlternate Rows and Columns:\n", alternate_rows_cols)
```

```python
last_2_rows_cols = matrix[-2: , -2:]
print("\nLast 2 Rows and Last 2 Columns:\n", last_2_rows_cols)
```

```python
employee_scores = np.array([
    [85, 90, 78, 92, 88],
    [70, 75, 80, 68, 72],
    [95, 88, 91, 84, 90]
])
print(employee_scores)
print("shape: " , employee_scores.shape)
```

```python
transposed_scores = employee_scores.T
print(transposed_scores)
print("shape: " , transposed_scores.shape)
```

```python
doubly_transposed_scores = transposed_scores.T
print(np.array_equal(employee_scores, doubly_transposed_scores))
```
